```python
# Importación de librerías

from matplotlib import pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from scipy import stats as st

# Configuración Estética Global
sns.set_theme(style="white", palette="muted")
plt.rcParams['figure.figsize'] = (12, 6)
plt.rcParams['axes.titlesize'] = 14
plt.rcParams['axes.titleweight'] = 'bold'

# Carga de datos
try:
    df = pd.read_csv("games.csv")
    df.columns = df.columns.str.lower()
    print("✅ Dataset cargado y columnas normalizadas.")
except FileNotFoundError:
    print("❌ Error: Archivo 'games.csv' no encontrado.")
```

### 🛠️ Fase 1: Data Engineering & Sanitization
En esta etapa inicial, se ejecuta un proceso de curaduría de datos enfocado en la **calidad analítica**. A diferencia de los enfoques convencionales, se evita la imputación masiva de valores ausentes con promedios globales para prevenir el "ruido estadístico" (por ejemplo, evitar la asignación errónea de lanzamientos de PS4 a años previos a su salida al mercado).

**Protocolo de Limpieza Aplicado:**

* **Tratamiento de Ambigüedades**: Se identifican y transforman los valores `TBD` (To Be Determined) en valores nulos reales para permitir un procesamiento numérico fluido.
* **Preservación de la Integridad**: Se opta por el uso de valores centinela (como `-1` o `unknown`) en lugar de la eliminación de registros. Esta decisión estratégica permite conservar el volumen de la muestra para análisis donde otras variables (como ventas o género) siguen siendo válidas y valiosas.
* **Sincronización de Tipos**: Ajuste de formatos temporales y categóricos para optimizar el consumo de memoria y la velocidad de ejecución de los algoritmos de agrupación posteriores.

```python
# Obtención y depuración de registros con valores nulos en dimensiones críticas
df = df.dropna(subset=['name', 'genre'])

# Configuración y normalización de variables categóricas (Strings a Lowercase)
# Estandarización para evitar duplicidad por discrepancias de escritura.
for col in ['name', 'genre', 'platform']:
    df[col] = df[col].str.lower()

# Definición de tipos de datos y gestión de valores 'TBD'
# Transformación de scores a formato numérico con coerción de errores para neutralizar ruido.
df['user_score'] = pd.to_numeric(df['user_score'], errors='coerce')

# Configuración de imputación técnica bajo parámetros de mejores prácticas
# Uso de valores centinela (-1) para evitar el sesgo por imputación de medias globales.
df['year_of_release'] = df['year_of_release'].fillna(-1).astype(int)
df['critic_score'] = df['critic_score'].fillna(-1).astype(int)
df['user_score'] = df['user_score'].fillna(-1)
df['rating'] = df['rating'].fillna("unknown")

# Definición de Métrica de Desempeño Global (KPI: Total Sales)
# Consolidación de ingresos transaccionales de todas las regiones geográficas.
df["total_sales"] = df[["na_sales", "eu_sales", "jp_sales", "other_sales"]].sum(axis=1)

# Obtención de perfil de datos sanitizado para auditoría visual
print("📊 Perfil de integridad de datos finalizado:")
display(df.sample(5))
```

### 📈 Fase 2: Análisis de Dinámica de Plataformas (Time Series Analysis)
Para garantizar la precisión del modelo predictivo de 2017, es imperativo identificar la **"Curva de Obsolescencia"** de los sistemas de hardware. Este análisis de series temporales no solo permite determinar el periodo de vida útil promedio de una consola, sino que fundamenta la delimitación del **periodo moderno (2014-2016)**.

**Objetivos del Diagnóstico:**

* **Determinación del Ciclo de Vida**: Identificar los patrones de ascenso, madurez y declive comercial para predecir qué plataformas actuales están en riesgo de obsolescencia terminal.
* **Filtrado de Ruido Estadístico**: Justificar la exclusión de hardware descontinuado que podría sesgar las proyecciones de ventas futuras.
* **Establecimiento de la Ventana Estratégica**: Definir el horizonte temporal relevante para capturar la dinámica actual del mercado, asegurando que el entrenamiento de datos sea coherente con las condiciones tecnológicas de 2017.

```python
# Obtención del Top 10 de plataformas con mayor volumen transaccional histórico
top_10_platforms = df.groupby('platform')['total_sales'].sum().sort_values(ascending=False).head(10).index

# Configuración de dataset para análisis de series temporales (Exclusión de años no identificados)
life_cycle_data = df[(df['platform'].isin(top_10_platforms)) & (df['year_of_release'] > 0)]

# Configuración del entorno visual
plt.figure(figsize=(14, 6))
sns.lineplot(data=life_cycle_data, x='year_of_release', y='total_sales', hue='platform', palette='viridis', marker='o')

# Definición de rótulos y parámetros de visualización
plt.title('Ciclo de Vida y Curva de Obsolescencia de Plataformas Líderes', pad=20)
plt.ylabel('Ventas Globales (Millones USD)')
plt.xlabel('Año de Lanzamiento')
plt.legend(title='Plataformas', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(axis='y', alpha=0.3)
plt.show()
```

#### 🔍 Hallazgos Clave: Dinámica del Ciclo de Vida (Longevidad Comercial)

El análisis de las series temporales para las plataformas líderes revela patrones críticos de comportamiento de mercado:

* **Periodo de Vida Útil**: Se identifica que las plataformas alcanzan su madurez y declive en un ciclo de aproximadamente **7 a 10 años**. Este hallazgo es fundamental para predecir la obsolescencia de los sistemas actuales.
* **Picos de Rentabilidad**: La mayoría de las plataformas experimentan su "Edad de Oro" (pico de ventas) entre el **año 3 y 5** después de su lanzamiento. 
* **Justificación de la Ventana Estratégica**: Observamos que para 2016, consolas históricas como PS3, Xbox 360 y Wii han completado su ciclo, mostrando ventas marginales. Esto **valida estadísticamente nuestra decisión de filtrar el dataset hacia el periodo 2014-2016**, asegurando que el modelo predictivo para 2017 se base únicamente en plataformas con tracción vigente.

> **Insight para Decision Making:** El foco estratégico debe desplazarse de las plataformas con mayor "Venta Histórica Total" hacia aquellas que muestran una **pendiente de crecimiento o estabilidad en el último trienio**, minimizando el riesgo de invertir en inventario de software para hardware en fase de retiro.

#### 🏆 Mix de Producto: Dominancia de Títulos por Ecosistema
Este análisis identifica los 10 videojuegos con mayor volumen de ventas en el periodo moderno. La visualización utiliza barras agrupadas con la **paleta de colores institucional**, permitiendo identificar de forma inmediata qué plataformas capturan la mayor cuota de mercado para cada título específico.

```python
# Obtención y transformación de matriz de ventas por título
pivot_games = df_modern.pivot_table(index='name', columns='platform', values='total_sales', aggfunc='sum').fillna(0)
top_10_names = pivot_games.sum(axis=1).nlargest(10).index
top_10_filtered = pivot_games.loc[top_10_names]

# Configuración del dataset (Melt)
df_top_10_long = top_10_filtered.reset_index().melt(id_vars='name', var_name='platform', value_name='sales')

# Gestión de consistencia cromática (Seguridad de Diccionario)
# Obtención de todas las plataformas presentes en este gráfico específico
present_platforms = df_top_10_long['platform'].unique()

# Configuración de colores de respaldo: 
# Si una plataforma no está en nuestro diccionario institucional, le asignamos gris.
current_palette = {p: platform_colors.get(p, '#d3d3d3') for p in present_platforms}

# Configuración del entorno visual
plt.figure(figsize=(15, 7))

# Ejecución de gráfico de barras agrupadas
sns.barplot(
    data=df_top_10_long, 
    x='name', 
    y='sales', 
    hue='platform', 
    palette=current_palette  # Usamos la paleta validada
)

# Configuración técnica de rótulos
plt.title('Top 10 Títulos Globales: Desglose de Ventas por Plataforma (2014-2016)', pad=20, fontsize=15)
plt.xlabel('Nombre del Videojuego')
plt.ylabel('Ventas (Millones USD)')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Plataformas', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(axis='y', linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()
```

La visualización del **Mix de Producto** permite diagnosticar la salud competitiva de las plataformas a través de sus títulos más exitosos:

* **Dominancia Multiplataforma**: Títulos líderes como *Call of Duty: Black Ops 3* y *Grand Theft Auto V* demuestran que la mayor parte del flujo de capital se concentra en el ecosistema **PS4 (Azul)** y **Xbox One (Verde)**. Este es un indicador clave de que, a pesar de tener un hardware menos potente, la propiedad intelectual exclusiva sigue siendo el motor de supervivencia de Nintendo frente a la competencia de alto rendimiento.
* **Larga Cola de Generación Anterior**: Se observa que éxitos masivos como *Minecraft* y *FIFA 15* mantienen barras visibles en plataformas "Legacy" como **PS3** y **Xbox 360**. ### 🗺️ Perfilado de Mercado: Adopción de Plataformas por Región
Para una estrategia de negocio óptima, es fundamental comprender las dinámicas de consumo a nivel regional. Este análisis visualiza las plataformas más exitosas en Norteamérica, Europa y Japón, permitiendo identificar **divergencias en la adopción de hardware** y la necesidad de una localización de inventario. El uso de la paleta de colores institucional garantiza la coherencia visual entre todos los segmentos geográficos.

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

for i, (col, title) in enumerate(regions.items()):
    # Obtención del Top 5 de plataformas por volumen de ventas regionales
    top_5 = df_modern.groupby('platform')[col].sum().sort_values(ascending=False).head(5)
    
    # Configuración de barplot con consistencia cromática de plataforma
    # Nota: Aplicación de lógica de fallback para colores previniendo aparición de plataforma no esperada
    present_platforms_in_region = top_5.index.unique()
    current_region_palette = {p: platform_colors.get(p, '#d3d3d3') for p in present_platforms_in_region}

    sns.barplot(
        x=top_5.index, 
        y=top_5.values, 
        ax=axes[i], 
        hue=top_5.index,          
        palette=current_region_palette,
        legend=False              
    )
    
    # Configuración de parámetros estéticos y rotulación técnica
    axes[i].set_title(f'Top 5 Plataformas en {title}', fontsize=14, pad=15)
    axes[i].set_ylabel('Ventas (Millones USD)')
    axes[i].set_xlabel('Plataforma')
    axes[i].grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

* **Dominancia Occidental (NA y EU)**: Se observa una clara hegemonía de **PS4 (Azul)** y **Xbox One (Verde)**, que lideran consistentemente los mercados de Norteamérica y Europa. Esto valida una estrategia unificada de comercialización y distribución para estas dos regiones, enfocándose en el hardware de octava generación.
* **Anomalía del Mercado Japonés**: La dinámica de Japón contrasta drásticamente. La **Nintendo 3DS (Rojo oscuro)** no solo domina, sino que excluye a la Xbox One del Top 5. Esto subraya la necesidad de una estrategia de mercado altamente localizada, priorizando consolas portátiles y títulos de nicho que resuenen con la cultura local.
* **Disparidad de Competencia**: La intensa competencia entre PS4 y Xbox One en Occidente se desvanece en Japón, donde la 3DS compite en un segmento diferente con una propuesta de valor única. Esto sugiere que las estrategias de precios y promociones deben ajustarse a la madurez y saturación de cada mercado.

### 🧪 Fase 5: Verificación de Hipótesis y Análisis Inferencial
Para transicionar de la observación visual a la certeza estadística, se aplica el rigor del **T-Test para muestras independientes**. El objetivo es validar si las tendencias detectadas en las fases previas son representativas de la población o si son fluctuaciones aleatorias en los datos.

**Protocolo de Validación:**

1.  **Hipótesis A (Paridad de Plataformas)**: Evaluación de la igualdad en las calificaciones de usuarios entre **Xbox One y PC**. Este análisis busca confirmar si existe un perfil de consumidor homogéneo para el ecosistema de Microsoft.
2.  **Hipótesis B (Diferenciación por Género)**: Comparación de los niveles de satisfacción entre los géneros de **Acción y Deportes**. Esta prueba es crítica para determinar si la inversión en inventario debe priorizar un género sobre otro basándose en la aceptación del usuario.

**Metodología Técnica:**
* **Prueba de Levene**: Aplicada previo al T-Test para verificar la igualdad de varianzas (homocedasticidad) y ajustar dinámicamente el modelo (Welch's T-Test si es necesario).
* **Nivel de Significancia ($\alpha$ = 0.05)**: Establecimiento de un umbral de confianza del 95% para rechazar o fallar en rechazar las hipótesis nulas.

```python
# Mantenemos tu función original que está muy bien estructurada
def check_hypotheses(sample1, sample2, alpha=0.05):
    # Definición de la prueba de Levene para verificación de homocedasticidad
    _, p_levene = st.levene(sample1, sample2)
    equal_var_setting = p_levene > alpha
    
    # Configuración y ejecución de T-Test (Ajuste de Welch dinámico)
    results = st.ttest_ind(sample1, sample2, equal_var=equal_var_setting)
    
    # Obtención de métricas de probabilidad
    print(f"P-Value obtenido: {results.pvalue:.10f}")
    if results.pvalue < alpha:
        return "❌ Conclusión: Rechazo de Hipótesis Nula (Diferencia significativa detectada)."
    else:
        return "✅ Conclusión: Fallo al rechazar Hipótesis Nula (Sin evidencia de diferencia)."

# --- EJECUCIÓN DINÁMICA ---

# Prueba 1: Xbox One vs PC
print("Validación A: Xbox One vs PC (User Score)")
sample_xone = df_modern[(df_modern['platform'] == 'xone') & (df_modern['user_score'] != -1)]['user_score']
sample_pc = df_modern[(df_modern['platform'] == 'pc') & (df_modern['user_score'] != -1)]['user_score']
print(check_hypotheses(sample_xone, sample_pc))

print("-" * 70) # Divisor estético

# Prueba 2: Acción vs Deportes
print("Validación B: Género Acción vs Género Deportes (User Score)")
sample_action = df_modern[(df_modern['genre'] == 'action') & (df_modern['user_score'] != -1)]['user_score']
sample_sports = df_modern[(df_modern['genre'] == 'sports') & (df_modern['user_score'] != -1)]['user_score']
print(check_hypotheses(sample_action, sample_sports))
```
